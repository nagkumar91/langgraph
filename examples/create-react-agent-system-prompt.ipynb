```python
import os
try:
    from langchain_azure_ai.callbacks.tracers import AzureAIOpenTelemetryTracer
    from langchain_core.callbacks.manager import add_global_callback
except ImportError:
    AzureAIOpenTelemetryTracer = None
    add_global_callback = None

_NOTEBOOK_TRACER = None
_ENABLE_CONTENT_RECORDING = os.getenv("AZURE_TRACING_GEN_AI_CONTENT_RECORDING_ENABLED", "true").lower() == "true"
if AzureAIOpenTelemetryTracer and add_global_callback:
    connection_string = os.getenv("APPLICATION_INSIGHTS_CONNECTION_STRING")
    if connection_string:
        tracer_name = os.getenv("AZURE_NOTEBOOK_TRACER_NAME", "langgraph-notebook")
        _NOTEBOOK_TRACER = AzureAIOpenTelemetryTracer(
            connection_string=connection_string,
            enable_content_recording=_ENABLE_CONTENT_RECORDING,
            name=tracer_name,
            id=tracer_name,
        )
        add_global_callback(_NOTEBOOK_TRACER)
        print(f"Azure tracer enabled: {tracer_name}")
    else:
        print("Azure tracer disabled: APPLICATION_INSIGHTS_CONNECTION_STRING not set")
else:
    print("Azure tracer unavailable; install langchain-azure-ai and langchain-core to enable.")
```
